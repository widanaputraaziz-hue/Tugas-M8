```python
#Case No. 1
pelamar = [("Budi", 3.2, 5), ("Siti", 3.6, 3), ("Dodi", 3.8, 1)]
print(f"{"="*10} HASIL PELAMAR ASISTEN {"="*10}")
for nama, ipk, semester in pelamar:
    if ipk >= 3.5 and semester >= 2:
        status = "DITERIMA"
    else:
        status = "TIDAK DITERIMA"

    print(f"Mahasiswa: {nama} | Status: {status}")
```

```python
#Case No. 2
karyawan = [("Joko", "IT", 7000), ("Rina", "HR", 5000), ("Toni", "Data", 8000), ("Lia", "IT", 6500)]
total = 0
print("=== GAJI KARYAWAN ===")
for nama, pekerjaan, gaji in karyawan:
    if pekerjaan == "IT" or pekerjaan == "Data":
       total += gaji
       print (f"[-] {nama} : {gaji}")
print(f"="*21)
print(f"Total : {total}")
```

```python
#Case No. 3
penjualan = [
    ("Kopi", 60, False),
    ("Teh", 30, False),
    ("Boba", 80, True)
    ]
jumlah_minim = 50
print ("=== SISTEM PERINGATAN PENJUALAN ===")
for nama, jumlah_penjualan, promo in penjualan:

    if jumlah_penjualan > jumlah_minim and not promo :
        print(f"INFO : Produk {nama} terjual keras sebanyak {jumlah_penjualan}!")
```
